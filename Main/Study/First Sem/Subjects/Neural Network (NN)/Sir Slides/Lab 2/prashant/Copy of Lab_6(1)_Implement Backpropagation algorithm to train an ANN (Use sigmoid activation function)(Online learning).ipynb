```python
# Implement Backpropagation algorithm to train an ANN of configuration 2x2x1 to achieve XOR function.
# (Use sigmoid activation function)It squashes the O/P in the range of (0,1).
# implement online method.
# Importing the necessary library for numerical operations
import numpy as np

#Initialize input(XOR)
input_values = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
target_outputs = np.array([0, 1, 1, 0])

# Initialize parameters
np.random.seed(1) #fix a randomly generated number(provide same random value) in every run. (i.e seed(1))
inputlayer_size = 2
hiddenlayer_size = 2
outputlayer_size = 1

#Initialize weights and bias
Weight_inputlayer_hiddenlayer = np.random.uniform(-1, 1, (hiddenlayer_size, inputlayer_size))
#assign random value from -1 to 1
#(hiddenlayer_size, inputlayer_size) = (2, 2) (for XOR).
#Connects the input layer (2 neurons) to the hidden layer (2 neurons).
#for example weight is assigned on this way.
#[[ 0.5, -0.1],  # Weight from input 1 to two hidden neurons
# [ 0.3,  0.8]]  # Weight from input 2 to two hidden neurons
bias_hiddenlayer = np.random.uniform(-1, 1, (hiddenlayer_size, 1))
#(hiddenlayer_size, 1) = (2,1)
#One bias per hidden neuron.
#[[0.2],  # Bias for hidden neuron 1
# [0.1]]  # Bias for hidden neuron 2 
Weight_hiddenlayer_outputlayer = np.random.uniform(-1, 1, (outputlayer_size, hiddenlayer_size))
bias_outputlayer = np.random.uniform(-1, 1, (outputlayer_size, 1))
print("weights:")
print(Weight_inputlayer_hiddenlayer)
print(Weight_hiddenlayer_outputlayer)
print("bias:")
print(bias_hiddenlayer)
print(bias_outputlayer)

#Why Initialize Randomly?
#Avoid Zero Initialization:
#----If all weights start as 0, neurons evolve identically (no learning).
#Small Random Values:
#----Prevents saturation of sigmoid (large values → gradients ≈ 0).
#Uniform Distribution:
#----Simple and effective for small networks.

#learning rate
learning_rate = 0.5
epochs = 10000
a=1

#Implement Backpropagation algorithm
#foward phase
for epoch in range(epochs):
    print(f"\nEpoch {epoch + 1}")

# Online learning - process each sample individually
    for i in range(len(input_values)):
# Get current input and target
        x = input_values[i].reshape(-1, 1)
        # Reshape input to column vector (2x1)
        # Gets the i-th input (e.g., [0, 0] at the begining )
        target = np.array([target_outputs[i]]).reshape(-1, 1)
        # Reshape target to column vector (1x1)
        # Gets the i-th input (e.g., [0] at the begining )

        # Forward pass to hidden layer
        # calculate the weighted sum with bias of input layer(z = w1*x1 + w2*x2 + bias)
        hidden_layer_weighted_sum_with_bias = np.dot(Weight_inputlayer_hiddenlayer, x) + bias_hiddenlayer
        # Apply sigmoid activation
        hidden_layer_output = 1 / (1 + np.exp(- a * hidden_layer_weighted_sum_with_bias))
        # Forward pass to output layer
        # calculate the weighted sum with bias of input layer(z = w1*x1 + w2*x2 + bias)
        output_layer_weighted_sum_with_bias = np.dot(Weight_hiddenlayer_outputlayer, hidden_layer_output) + bias_outputlayer
        # Apply sigmoid activation
        predicted_output = 1 / (1 + np.exp(- a * output_layer_weighted_sum_with_bias))

        # Calculate error (Mean Squared Error)
        error = 0.5 * (target - predicted_output) ** 2
        print(f"Input: {x.T}, Output: {predicted_output[0,0]:.4f}, Target: {target[0,0]}")

        # Backward pass (Backpropagation)
        # Calculate output layer error gradient
        output_layer_error_gradient = (target - predicted_output) * a * predicted_output * (1 - predicted_output)
        # Calculate hidden layer error gradient
        hidden_layer_error_gradient = np.dot(Weight_hiddenlayer_outputlayer.T, output_layer_error_gradient) * a * hidden_layer_output * (1 - hidden_layer_output)

        # Update weights and biases (Online learning - update after each sample)
        # Update output layer weights
        Weight_hiddenlayer_outputlayer += learning_rate * np.dot(output_layer_error_gradient, hidden_layer_output.T)
        # Update output layer bias
        bias_outputlayer += learning_rate * output_layer_error_gradient
        # Update hidden layer weights
        Weight_inputlayer_hiddenlayer += learning_rate * np.dot(hidden_layer_error_gradient, x.T)
        # Update hidden layer bias
        bias_hiddenlayer += learning_rate * hidden_layer_error_gradient


# After training, test the network
print("\nTesting the trained network:")
for i in range(len(input_values)):
    x = input_values[i].reshape(-1, 1)

    # Forward pass only
    hidden_layer_weighted_sum_with_bias = np.dot(Weight_inputlayer_hiddenlayer, x) + bias_hiddenlayer
    hidden_layer_output = 1 / (1 + np.exp(- a * hidden_layer_weighted_sum_with_bias))
    output_layer_weighted_sum_with_bias = np.dot(Weight_hiddenlayer_outputlayer, hidden_layer_output) + bias_outputlayer
    predicted_output = 1 / (1 + np.exp(- a * output_layer_weighted_sum_with_bias))

    # Round to nearest integer (0 or 1)
    predicted_class = 1 if predicted_output > 0.5 else 0
    print(f"Input: {x.T}, Output: {predicted_output[0,0]:.4f}, Predicted: {predicted_class}, Target_output: {target_outputs[i]}")







```
